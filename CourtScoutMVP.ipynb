**Task 4 - Classifying Each Shot Attempt into designated zones**

```python
print(merged_data['BASIC_ZONE'].unique())
countedShots = merged_data['BASIC_ZONE'].value_counts()
print('\n\n\n\n\n', countedShots)

shots_mapped = {
    'Above the Break 3' : 'Above-The-Break 3',
    'Restricted Area' : 'Restricted Area',
    'In The Paint (Non-RA)' : 'Paint (Non-Restricted)',
    'Mid-Range' : 'Mid-Range',
    'Right Corner 3' : 'Corner 3',
    'Left Corner 3' : 'Corner 3'
}

merged_data = merged_data.dropna(subset=['BASIC_ZONE'])
merged_data['COURTSCOUT_ZONE'] = merged_data['BASIC_ZONE'].map(shots_mapped)
print('\n\n\n\n\n\n\n', merged_data)

missingValues = merged_data['COURTSCOUT_ZONE'].isnull().sum()
print('\n\n\n\n', missingValues)

print(merged_data['COURTSCOUT_ZONE'].value_counts())

print(merged_data[["PLAYER_NAME", "GAMES_NUMBER", "BASIC_ZONE", "COURTSCOUT_ZONE"]].head(10))
```

**STEP 5 - Select & Validate A Target Game**

```python
# Step 1 - Configure the first recent shooting profile
target_game = 82
target_game_index_one = target_game - 5
target_game_index_two = target_game - 1
recent_five_game_statistics = merged_data[merged_data['GAMES_NUMBER'].between(target_game_index_one, target_game_index_two)]
print(recent_five_game_statistics)

# Step 2 - Configure the second shooting brief statistics
season_to_date = merged_data[merged_data['GAMES_NUMBER'] < target_game]
print(season_to_date)


# Step 3 - Contain Shots from the target game
actual_target = merged_data[merged_data['GAMES_NUMBER'] == target_game]
print(actual_target)

print(recent_five_game_statistics['GAMES_NUMBER'].unique())
print(season_to_date['GAMES_NUMBER'].unique())
print(actual_target['GAMES_NUMBER'].unique())
```

**STEP 6 - Generate the Shooting Profiles**

```python
# Step 1 - The Recent Statistics
recent_total_shot_attempts = recent_five_game_statistics.groupby('COURTSCOUT_ZONE')['SHOT_MADE'].size()
print('\n\n---------------------------\n SHOTS ATTEMPTED IN THE LAST 5 GAMES \n---------------------------\n', recent_total_shot_attempts)

# Step 1.1 Calculate Shots Made in the last 5 Games
print('\n\n---------------------------\n SHOT MAKES LAST 5 GAMES \n---------------------------\n')
recent_total_shot_makes = recent_five_game_statistics.groupby('COURTSCOUT_ZONE')['SHOT_MADE'].sum()
print(recent_total_shot_makes)

# Step 1.2 Calculate % of Shots Converted
recent_shot_attempts_converted = (recent_total_shot_makes / recent_total_shot_attempts) * 100
recent_shot_attempts_converted = recent_shot_attempts_converted.round(2)
print('\n\n---------------------------\n % OF SHOTS CONVERTED \n---------------------------\n', recent_shot_attempts_converted)

# Step 1.3 Calculate The Frequency Of Each Shot defined in each CourtScout Zone
recent_shots_distribution = recent_five_game_statistics.groupby('COURTSCOUT_ZONE')['SHOT_MADE']
recent_shots_distribution = recent_shots_distribution.agg(["size", "sum"])
recent_shot_total =  recent_total_shot_attempts.sum()
recent_shot_total = (recent_total_shot_attempts / recent_shot_total) * 100
recent_shot_total = recent_shot_total.round(2)
print('\n\n---------------------------\n FREQUENCY SHOT DISTRIBUTION - LAST 5 GAMES (AS A %)\n---------------------------\n', recent_shot_total)

# Step 1.4 Output the CourtScout Shot Distribution Chart (FGA, FGM, FG%, Frequency %)
recent_shots_distribution = recent_shots_distribution.rename(columns={'size' : 'FGA', 'sum' : 'FGM'})
recent_shots_distribution['FG%'] = recent_shot_attempts_converted
recent_shots_distribution['Frequency %'] = recent_shot_total
print('\n\n---------------------------\n RECENT 5 GAME SHOOTING BRIEF\n---------------------------\n', recent_shots_distribution)

print('\n\n\n\n\n\n---------------------------\n SEASON TO DATE STATISTICS\n---------------------------\n')












# Step 2 - The Season-to-date statistics

# Step 2.1 - Calculate Total Shots Attempted From Game 1 - Target Game
season_to_date_shot_attempts = season_to_date.groupby('COURTSCOUT_ZONE')['SHOT_MADE'].size()
print('\n\n---------------------------\n SEASON-TO-DATE SHOT ATTEMPTS \n---------------------------\n', season_to_date_shot_attempts)

# Step 2.2 - Calculate Total Shots Made From Game 1 - One Game Before Target Game
season_to_date_shots_made = season_to_date.groupby('COURTSCOUT_ZONE')['SHOT_MADE'].sum()
print('\n\n---------------------------\n SEASON-TO-DATE SHOTS MADE \n---------------------------\n', season_to_date_shots_made)

# Step 2.3 - Calculate Season-To-Date FG% From Game 1 - One Game Before Target Game
season_to_date_field_goal_percentage = (season_to_date_shots_made / season_to_date_shot_attempts) * 100
season_to_date_field_goal_percentage = season_to_date_field_goal_percentage.round(2)

# Step 2.4 - Calculate Season-To-Date Shot Distribution in Each CourtScout Zone From Game 1 - One Game Before Target Game
season_to_date_shots_distribution = season_to_date.groupby('COURTSCOUT_ZONE')['SHOT_MADE']
season_to_date_shots_distribution = season_to_date_shots_distribution.agg(["size", "sum"])
season_to_date_shot_frequency = season_to_date_shot_attempts.sum()
season_to_date_shot_frequency = (season_to_date_shot_attempts / season_to_date_shot_frequency) * 100
print('\n\n\n\n\n', season_to_date_shot_frequency.round(2))

# Step 2.5 - Define the Season-To-Date Shot Distribution
season_to_date_shots_distribution = season_to_date_shots_distribution.rename(columns={'size' : 'FGA', 'sum' : 'FGM'})
season_to_date_shots_distribution['FG%'] = season_to_date_field_goal_percentage
season_to_date_shots_distribution['Frequency %'] = season_to_date_shot_frequency.round(2)

# Step 2.5 - Print Out The Season-To-Date Statistics
print('\n\n---------------------------\n SEASON-TO-DATE GAME SHOT DISTRIBUTION \n---------------------------\n', season_to_date_shots_distribution)




print('\n\n\n\n\n\n\n\n\n\n\n\n\n\n\n\n')





# Step 3 - Generate Actual Target Report Game

# Step 3.1 - Calculate Total Shots Attempted in Target Game
actual_target_game_shot_attempts = actual_target.groupby('COURTSCOUT_ZONE')['SHOT_MADE'].size()
print('\n--------------------------------\nACTUAL SHOTS MADE\n--------------------------------\n', actual_target_game_shot_attempts)

# Step 3.2 - Calculate Total Shots Made In Target Game
print('\n\n---------------------------\n SHOT MAKES TARGET GAME \n---------------------------\n')
actual_target_game_shot_makes = actual_target.groupby('COURTSCOUT_ZONE')['SHOT_MADE'].sum()
print(actual_target_game_shot_makes)

# Step 3.3 - Calculate the % Of Shots Converted
print('\n\n---------------------------\n % OF SHOTS CONVERTED \n---------------------------\n')
actual_target_game_shots_converted = (actual_target_game_shot_makes / actual_target_game_shot_attempts) * 100
print(actual_target_game_shots_converted.round(2))

# Step 3.4 - Calculate the Frequency Of Each Shot Defined in Each CourtScout Zone
target_game_shot_distribution = actual_target.groupby('COURTSCOUT_ZONE')['SHOT_MADE']
target_game_shot_distribution = target_game_shot_distribution.agg(["size", "sum"])
target_game_shot_total = actual_target_game_shot_attempts.sum()
target_game_shot_total = (actual_target_game_shot_attempts / target_game_shot_total) * 100
target_game_shot_total = target_game_shot_total.round(2)
print('\n\n---------------------------\n FREQUENCY SHOT DISTRIBUTION - TARGET GAME (AS A %)\n---------------------------\n', target_game_shot_total)

# Step 3.5 - Output the CourtScout Shot Distribution Chart (FGA, FGM, FG%, Frequency %)
target_game_shot_distribution = target_game_shot_distribution.rename(columns={'size' : 'FGA', 'sum' : 'FGM'})
target_game_shot_distribution['FG%'] = actual_target_game_shots_converted
target_game_shot_distribution['Frequency %'] = target_game_shot_total

print('\n\n---------------------------\n TARGET GAME SHOOTING BRIEF\n---------------------------\n', target_game_shot_distribution)
```
